```python
import sys
from types import ModuleType

# Create a mock grid_env_det module
mock_env = ModuleType('grid_env_det')

def mock_step(state, action):
    # A simple deterministic transition logic for demonstration
    # Modify this dictionary to define custom transitions, rewards, and terminal states
    transitions = {
        ("S0", "right"): ("S1", 1.0, False),
        ("S0", "up"): ("S3", 0.0, False),
        ("S1", "down"): ("S2", 2.0, True),
        ("S3", "right"): ("S4", 0.0, False),
        ("S4", "up"): ("S5", 5.0, True)
    }

    key = (state, action)
    if key in transitions:
        return transitions[key]
    else:
        # Default fallback transition
        return (state, 0.0, True)

mock_env.step = mock_step
sys.modules['grid_env_det'] = mock_env
print("Mock grid_env_det module created and registered!")
```
