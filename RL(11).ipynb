```python

gamma = 0.9
theta = 0.000001

states = ["S0", "S1"]
actions = ["up", "down", "left", "right"]

transitions = {
    "S0": {
        "up": ("S0", -1),
        "down": ("S1", -1),
        "left": ("S0", -1),
        "right": ("S0", -1)
    },
    "S1": {
        "up": ("S0", -1),
        "down": ("S1", -1),
        "left": ("S1", -1),
        "right": ("G", 100)
    }
}

V = {"S0": 0, "S1": 0, "G": 0}

policy = {
    "S0": "up",
    "S1": "up"
}

print("GRID WORLD MDP")
print("States:", states)
print("Actions:", actions)
print("Discount Factor:", gamma)

print("\nTRANSITIONS")

for state in states:
    for action in actions:
        next_state, reward = transitions[state][action]
        print(state, action, "->", next_state, "Reward:", reward)

iteration = 0

while True:
    print("\nPOLICY ITERATION:", iteration + 1)
    print("Current Policy:", policy)

    while True:
        new_V = V.copy()

        for state in states:
            action = policy[state]
            next_state, reward = transitions[state][action]
            new_V[state] = reward + gamma * V[next_state]

        delta = max(abs(new_V[s] - V[s]) for s in states)

        V = new_V

        print("State Values:", {s: round(V[s], 6) for s in states})
        print("Delta:", round(delta, 8))

        if delta < theta:
            break

    Q = {}

    for state in states:
        Q[state] = {}

        for action in actions:
            next_state, reward = transitions[state][action]
            Q[state][action] = reward + gamma * V[next_state]

    print("\nACTION VALUES")

    for state in states:
        print(state, Q[state])

    new_policy = {}

    for state in states:
        best_action = max(Q[state], key=Q[state].get)
        new_policy[state] = best_action

    print("\nIMPROVED POLICY:", new_policy)

    iteration += 1

    if new_policy == policy:
        break

    policy = new_policy

print("\nFINAL OPTIMAL STATE VALUES")

for state in states:
    print(state, "=", round(V[state], 2))

print("G = 0 (Terminal)")

print("\nFINAL OPTIMAL POLICY")

for state in states:
    print(state, "->", policy[state])

print("\nFINAL ACTION VALUES")

for state in states:
    for action in actions:
        print(state, action, "=", round(Q[state][action], 2))

print("\nBELLMAN OPTIMALITY VERIFICATION")

for state in states:
    max_q = max(Q[state].values())

    print(state, "V* =", round(V[state], 2),
          "Max Q* =", round(max_q, 2),
          "Verified =", abs(V[state] - max_q) < 0.000001)
```
