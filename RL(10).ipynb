```python

gamma = 0.9
theta = 0.000001

states = ["S0", "S1"]
actions = ["up", "down", "left", "right"]

transitions = {
    "S0": {
        "up": ("Wall", -1),
        "down": ("S1", -1),
        "left": ("Wall", -1),
        "right": ("Wall", -1)
    },
    "S1": {
        "up": ("S0", -1),
        "down": ("Wall", -1),
        "left": ("Wall", -1),
        "right": ("G", 10)
    }
}

print("GRID WORLD MDP")
print("States:", states)
print("Actions:", actions)
print("Discount Factor:", gamma)
print("Normal Reward: -1")
print("Wall Reward: -1")
print("Goal Reward: +10")

print("\nTRANSITION VERIFICATION")

for state in states:
    for action in actions:
        next_state, reward = transitions[state][action]
        print(state, action, "->", next_state, "Reward:", reward)

V = {"S0": 0, "S1": 0, "G": 0, "Wall": 0}

print("\nINITIAL STATE VALUES:", V)

print("\nVALUE ITERATION")

iteration = 0

while True:
    new_V = {}

    for state in states:
        values = []

        for action in actions:
            next_state, reward = transitions[state][action]
            value = reward + gamma * V[next_state]
            values.append(value)

        new_V[state] = max(values)

    delta = max(abs(new_V[s] - V[s]) for s in states)

    V.update(new_V)
    iteration += 1

    print("Iteration", iteration, "V:", new_V, "Delta:", round(delta, 6))

    if delta < theta:
        break

print("\nOPTIMAL STATE VALUES")

for state in states:
    print(state, "=", round(V[state], 2))

print("\nACTION VALUES")

Q = {}

for state in states:
    Q[state] = {}

    for action in actions:
        next_state, reward = transitions[state][action]
        Q[state][action] = reward + gamma * V[next_state]

        print("Q(", state, ",", action, ") =",
              round(Q[state][action], 2))

print("\nOPTIMAL POLICY")

for state in states:
    maximum = max(Q[state].values())
    best_actions = [
        action for action in actions
        if abs(Q[state][action] - maximum) < 0.0000001
    ]

    print(state, "->", best_actions)

print("\nBELLMAN OPTIMALITY VERIFICATION")

for state in states:
    max_q = max(Q[state].values())

    print(state, ": V* =", round(V[state], 2),
          "Max Q* =", round(max_q, 2),
          "Verified =", abs(V[state] - max_q) < 0.000001)
```
